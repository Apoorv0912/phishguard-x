```python
import time
import socket
from urllib.parse import urlparse

url = legiurl["URLs"].iloc[0]
domain = urlparse(url).netloc.split(":")[0]

print("URL:", url)
print("Domain:", domain)

# DNS
start = time.time()
try:
    ip = socket.gethostbyname(domain)
    print(f"DNS: SUCCESS | {time.time() - start:.2f}s | IP: {ip}")
except Exception as e:
    print(f"DNS: FAILED | {time.time() - start:.2f}s | {type(e).__name__}")

# WHOIS
start = time.time()
try:
    whois_result = uf.get_whois_cached(domain)
    print(f"WHOIS: {'SUCCESS' if whois_result else 'EMPTY'} | {time.time() - start:.2f}s")
except Exception as e:
    print(f"WHOIS: FAILED | {time.time() - start:.2f}s | {type(e).__name__}")

# HTTP
start = time.time()
try:
    response = uf.requests.get(
        url,
        timeout=5,
        headers={"User-Agent": "Mozilla/5.0"},
        allow_redirects=True
    )
    print(f"HTTP: SUCCESS | {time.time() - start:.2f}s | Status: {response.status_code}")
except Exception as e:
    print(f"HTTP: FAILED | {time.time() - start:.2f}s | {type(e).__name__}")
```

```python
import pandas as pd
import time
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path

OUTPUT_FILE = Path("../data/processed/legitimate_features.csv")
CHECKPOINT_FILE = Path("../data/processed/checkpoints/legitimate_features_checkpoint.csv")

urls = legiurl["URLs"].tolist()

def extract_one(url):
    try:
        result = uf.featureExtraction(url, 0)

        if result is not None and len(result) == len(uf.feature_names):
            return result

        return None

    except Exception:
        return None


all_results = []
start_time = time.time()

BATCH_SIZE = 50
MAX_WORKERS = 5

for batch_start in range(0, len(urls), BATCH_SIZE):

    batch_urls = urls[batch_start:batch_start + BATCH_SIZE]

    batch_results = []

    with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:

        futures = [
            executor.submit(extract_one, url)
            for url in batch_urls
        ]

        for future in as_completed(futures):

            result = future.result()

            if result is not None:
                batch_results.append(result)

    all_results.extend(batch_results)

    completed = min(batch_start + BATCH_SIZE, len(urls))

    elapsed = time.time() - start_time

    print(
        f"Completed: {completed}/{len(urls)} | "
        f"Successful: {len(all_results)} | "
        f"Time: {elapsed / 60:.2f} min",
        flush=True
    )

    # Checkpoint after every 50 URLs
    pd.DataFrame(
        all_results,
        columns=uf.feature_names
    ).to_csv(CHECKPOINT_FILE, index=False)


# Final dataframe
legitimate_features = pd.DataFrame(
    all_results,
    columns=uf.feature_names
)

# Final save
legitimate_features.to_csv(
    OUTPUT_FILE,
    index=False
)

elapsed = time.time() - start_time

print("\n========== LEGITIMATE EXTRACTION COMPLETE ==========")
print(f"Successful: {len(all_results)}/{len(urls)}")
print(f"Failed: {len(urls) - len(all_results)}")
print(f"Time: {elapsed / 60:.2f} min")
print(f"Shape: {legitimate_features.shape}")
print(f"Columns: {len(legitimate_features.columns)}")
print(f"Saved: {OUTPUT_FILE}")
```

```python
from pathlib import Path
import pandas as pd
import time
from concurrent.futures import ThreadPoolExecutor, as_completed

OUTPUT_FILE = Path("../data/processed/phishing_features.csv")
CHECKPOINT_FILE = Path(
    "../data/processed/checkpoints/phishing_features_checkpoint.csv"
)

urls = phishurl["URLs"].tolist()

def extract_one(item):
    idx, url = item

    try:
        result = uf.featureExtraction(url, 1)

        if result is not None and len(result) == len(uf.feature_names):
            return idx, result

        return idx, None

    except Exception:
        return idx, None


all_results = {}
start_time = time.time()

MAX_WORKERS = 5
CHECKPOINT_EVERY = 50

with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:

    futures = [
        executor.submit(extract_one, item)
        for item in enumerate(urls)
    ]

    completed = 0

    for future in as_completed(futures):

        idx, result = future.result()
        completed += 1

        if result is not None:
            all_results[idx] = result

        print(
            f"Completed: {completed}/{len(urls)} | "
            f"Successful: {len(all_results)}",
            flush=True
        )

        if completed % CHECKPOINT_EVERY == 0:

            ordered_results = [
                all_results[i]
                for i in sorted(all_results)
            ]

            checkpoint_df = pd.DataFrame(
                ordered_results,
                columns=uf.feature_names
            )

            checkpoint_df.to_csv(
                CHECKPOINT_FILE,
                index=False
            )

            print(
                f"Checkpoint updated: {len(checkpoint_df)} rows",
                flush=True
            )


# Final CSV
ordered_results = [
    all_results[i]
    for i in sorted(all_results)
]

phish_features = pd.DataFrame(
    ordered_results,
    columns=uf.feature_names
)

phish_features.to_csv(
    OUTPUT_FILE,
    index=False
)

elapsed = time.time() - start_time

print("\n========== PHISHING EXTRACTION COMPLETE ==========")
print("Final shape:", phish_features.shape)
print("Successful:", len(phish_features))
print(f"Time: {elapsed / 60:.2f} min")
print("Saved:", OUTPUT_FILE)
```

```python
from pathlib import Path
import pandas as pd

LEGIT_FILE = Path("../data/processed/legitimate_features.csv")
PHISH_FILE = Path("../data/processed/phishing_features.csv")
FINAL_FILE = Path("../data/processed/phishguard_features.csv")

legit_features = pd.read_csv(LEGIT_FILE)
phish_features = pd.read_csv(PHISH_FILE)

# Verify both datasets before merging
assert legit_features.shape == (5000, 18)
assert phish_features.shape == (5000, 18)
assert list(legit_features.columns) == list(phish_features.columns)

# Merge
phishguard_features = pd.concat(
    [legit_features, phish_features],
    ignore_index=True
)

# Save fresh final dataset
phishguard_features.to_csv(FINAL_FILE, index=False)

print("Final dataset shape:", phishguard_features.shape)
print("\nLabel distribution:")
print(phishguard_features["Label"].value_counts().sort_index())

print("\nMissing values:", phishguard_features.isna().sum().sum())
print("Saved:", FINAL_FILE)
```
